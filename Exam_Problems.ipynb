# 📌 Problem 1 — Separable ODE via Substitution <a id='p1'></a>


## The Problem

Solve the ordinary differential equation:

$$((t-2)t - 1)\,x(t) = (t^2 - 1)\,\dot{x}(t)$$

Find the **general real solution** $x(t)$.


## 📖 Concept Recap: First-Order ODEs

### Separable ODE
An ODE of the form $\dot{x} = f(t)\,g(x)$ is **separable** — rearrange as:
$$\frac{dx}{g(x)} = f(t)\,dt$$
then integrate both sides.

### Substitution technique
If the ODE doesn't separate directly, look for a substitution $y = \phi(t, x)$ that simplifies it.

**Key idea:** if you can spot that both sides are parts of a known derivative, use that.  
Here the right-hand side $2tx + (t^2-1)\dot{x}$ is exactly $\frac{d}{dt}[(t^2-1)x]$ by the product rule:
$$\frac{d}{dt}[(t^2-1)x] = 2t\,x + (t^2-1)\dot{x}$$

### Solving $\dot{y} = y$
This is the simplest ODE.  Solution: $y(t) = Ce^t$.

### SymPy tools used
```python
t = symbols('t')
x = Function('x')
ode = Eq(lhs, rhs)
dsolve(ode, x(t))          # solves the ODE symbolically
```


## ✏️ Hand-Solution Walkthrough

**Step 1 — Expand the left side:**
$$((t-2)t - 1)x = (t^2 - 2t - 1)x = (t^2-1)x - 2tx$$

**Step 2 — Rearrange:**
$$(t^2-1)x - 2tx = (t^2-1)\dot{x}$$
$$(t^2-1)x = 2tx + (t^2-1)\dot{x}$$

**Step 3 — Spot the substitution:**  
The right side is exactly $\frac{d}{dt}[(t^2-1)x]$.  Set $y = (t^2-1)x$, so $\dot{y} = 2tx + (t^2-1)\dot{x}$.

**Step 4 — Simplified ODE:**
$$y = \dot{y}$$

**Step 5 — Solve:** General solution $y(t) = Ce^t$.

**Step 6 — Undo substitution:** $y = (t^2-1)x$, so:
$$\boxed{x(t) = \frac{Ce^t}{t^2 - 1}}$$


## 🐍 SymPy Solution

````python
t = symbols('t', real=True)
x = Function('x')

# ── Step 1: Define the ODE ────────────────────────────────────────────────
lhs = ((t - 2)*t - 1) * x(t)
rhs = (t**2 - 1) * x(t).diff(t)

ode = Eq(lhs, rhs)
print("ODE:")
pprint(ode)

# ── Step 2: Solve directly with dsolve ────────────────────────────────────
sol = dsolve(ode, x(t))
print("\nGeneral solution:")
pprint(sol)

# ── Step 3: Verify — substitute back and check both sides are equal ───────
x_sol = sol.rhs
xdot  = diff(x_sol, t)

lhs_check = ((t - 2)*t - 1) * x_sol
rhs_check = (t**2 - 1) * xdot
diff_check = simplify(lhs_check - rhs_check)
print("\nVerification (lhs - rhs should be 0):", diff_check)

# ── Step 4: Show the substitution approach manually ───────────────────────
print("\n── Manual substitution approach ──")
y = Function('y')
# y = (t^2 - 1)*x  =>  x = y/(t^2 - 1)
# dy/dt = 2t*x + (t^2-1)*xdot  =>  original ODE becomes dy/dt = y
ode_y = Eq(y(t).diff(t), y(t))
print("Reduced ODE after substitution:"); pprint(ode_y)
sol_y = dsolve(ode_y, y(t))
print("Solution for y:"); pprint(sol_y)
print("\nUndo substitution x = y/(t²-1):")
x_from_y = sol_y.rhs / (t**2 - 1)
pprint(Eq(x(t), simplify(x_from_y)))
````

# 📌 Problem 2 — N-Good Cobb-Douglas Utility Maximisation <a id='p2'></a>


## The Problem

A consumer has utility function:
$$u(x_1,\ldots,x_N) = \prod_{n=1}^N x_n^{\alpha_n}$$

with $\alpha_n > 0$, $\sum_{n=1}^N \alpha_n = 1$, wealth $w > 0$, prices $p_n > 0$.

**Question:** What share of wealth $w$ does the consumer spend on the **most expensive good**?


## 📖 Concept Recap: Lagrange Multipliers

### The constrained optimisation problem
$$\max_{x} f(x) \quad \text{subject to} \quad g(x) = c$$

### Lagrangian
$$\mathcal{L}(x, \lambda) = f(x) - \lambda(g(x) - c)$$

### First-Order Conditions (FOCs)
$$\frac{\partial \mathcal{L}}{\partial x_n} = 0 \quad\text{for all }n, \qquad g(x) = c$$

### Log-trick for products
Since $\log$ is monotone, $\max u(x) \Leftrightarrow \max \log u(x)$.  
For a product: $\log\prod x_n^{\alpha_n} = \sum \alpha_n \log x_n$ — turns products into sums.

### SymPy tools used
```python
symbols(...)                # declare N symbols at once
diff(L, x_n)                # partial derivative
solve([foc1, foc2, ..., budget], [x1, ..., xN, lam])
```


## ✏️ Hand-Solution Walkthrough

**Step 1 — Log-transform:** Maximise $\sum_{n=1}^N \alpha_n \ln x_n$ (same optimum).

**Step 2 — Lagrangian:**
$$\mathcal{L} = \sum_{n=1}^N \alpha_n \ln x_n - \lambda\Bigl(\sum_{n=1}^N p_n x_n - w\Bigr)$$

**Step 3 — FOC for good $n$:**
$$\frac{\partial \mathcal{L}}{\partial x_n} = \frac{\alpha_n}{x_n} - \lambda p_n = 0 \implies \alpha_n = \lambda p_n x_n$$

**Step 4 — Ratio of FOCs for goods $n$ and $k$:**
$$\frac{\alpha_n}{\alpha_k} = \frac{p_n x_n}{p_k x_k} \implies p_n x_n = \frac{\alpha_n}{\alpha_k} p_k x_k$$

**Step 5 — Sum over all $n$ and use budget:**
$$\sum_n p_n x_n = w \implies \frac{p_k x_k}{\alpha_k} \sum_n \alpha_n = w \implies p_k x_k = \alpha_k w$$

**Conclusion:** The expenditure share on **any** good $k$ is $\alpha_k$ — regardless of prices.  
So the most expensive good also gets share $\alpha_{\text{most expensive}}$.


## 🐍 SymPy Solution

````python
# ── Concrete N=3 case to verify ───────────────────────────────────────────
N = 3
x1, x2, x3, lam, w = symbols('x1 x2 x3 lambda w', positive=True)
p1, p2, p3          = symbols('p1 p2 p3', positive=True)
a1, a2, a3          = Rational(1,2), Rational(1,3), Rational(1,6)   # sum = 1

print(f"Shares: α1={a1}, α2={a2}, α3={a3}, sum={a1+a2+a3}")

# Log-utility (monotone transform)
log_u = a1*log(x1) + a2*log(x2) + a3*log(x3)

# Lagrangian
L = log_u - lam*(p1*x1 + p2*x2 + p3*x3 - w)

# FOCs
focs = [diff(L, x1), diff(L, x2), diff(L, x3), diff(L, lam)]
print("\nFOCs:"); [pprint(f) for f in focs]

# Solve the system
sol = solve(focs, [x1, x2, x3, lam], dict=True)[0]
print("\nMarshallian demands:")
for var, expr in sol.items():
    pprint(Eq(var, simplify(expr)))

# ── Compute expenditure shares ────────────────────────────────────────────
print("\nExpenditure shares  p_n * x_n* / w:")
shares = {
    'good 1': simplify(p1 * sol[x1] / w),
    'good 2': simplify(p2 * sol[x2] / w),
    'good 3': simplify(p3 * sol[x3] / w),
}
for good, share in shares.items():
    print(f"  {good}: {share}")

print("\n✅ Each share equals the corresponding α — independent of prices!")

# ── General symbolic proof for N=2 ───────────────────────────────────────
print("\n── General 2-good symbolic proof ──")
x, y, lam2, p, q, W, a, b = symbols('x y lambda p q W a b', positive=True)
# constraint: a + b = 1  →  use b = 1-a
L2 = a*log(x) + (1-a)*log(y) - lam2*(p*x + q*y - W)
sol2 = solve([diff(L2,x), diff(L2,y), diff(L2,lam2)], [x,y,lam2], dict=True)[0]
print("Marshallian demand x*:"); pprint(simplify(sol2[x]))
print("Expenditure share p*x*/W:"); pprint(simplify(p*sol2[x]/W))
````

# 📌 Problem 3 — Linear ODE System: Long-Run Behaviour <a id='p3'></a>


## The Problem

Given the system:
$$\dot{x} = \frac{x}{2} + 3y, \qquad \dot{y} = \frac{3x}{4} + \frac{y}{2}$$

Let $(\bar{x}(t), \bar{y}(t))$ be the solution with initial condition $(\bar{x}(0), \bar{y}(0)) = (1,1)$.

**Question:** What is $\lim_{t \to \infty} \bar{x}(t)/\bar{y}(t)$?


## 📖 Concept Recap: Linear ODE Systems

### The System $\dot{\mathbf{x}} = A\mathbf{x}$
General solution: $\mathbf{x}(t) = \sum_i C_i \mathbf{v}_i e^{\lambda_i t}$  
where $\lambda_i$ are eigenvalues and $\mathbf{v}_i$ are corresponding eigenvectors.

### Stability Classification (for 2×2)
| $\text{tr}(A)$ | $\det(A)$ | Type |
|--------------|----------|------|
| $< 0$ | $> 0$ | Stable node / spiral → $\mathbf{x} \to 0$ |
| $> 0$ | $> 0$ | Unstable node / spiral → $\|\mathbf{x}\| \to \infty$ |
| any | $< 0$ | Saddle — one direction grows, one shrinks |

### Long-run direction
If $\lambda_1 > \lambda_2$ (both real), the $e^{\lambda_1 t}$ term dominates as $t \to \infty$.  
The solution aligns with eigenvector $\mathbf{v}_1$, so the ratio of components → the ratio in $\mathbf{v}_1$.

**Exception:** if initial condition lies **exactly** on the stable subspace (spanned by $\mathbf{v}_2$),  
the solution decays along $\mathbf{v}_2$ instead.

### SymPy tools used
```python
A.eigenvals()        # eigenvalues
A.eigenvects()       # eigenvectors  
A.diagonalize()      # P, D such that A = P D P^{-1}
dsolve(system, ...)  # solve ODE system directly
```


## ✏️ Hand-Solution Walkthrough

**Step 1 — Matrix:**
$$A = \begin{pmatrix} 1/2 & 3 \\ 3/4 & 1/2 \end{pmatrix}$$

**Step 2 — Characteristic polynomial:**
$$\det(A - \lambda I) = \left(\frac{1}{2} - \lambda\right)^2 - \frac{9}{4} = \lambda^2 - \lambda - 2 = (\lambda-2)(\lambda+1)$$

**Step 3 — Eigenvalues:** $\lambda_1 = 2$, $\lambda_2 = -1$.

**Step 4 — Eigenvectors:**
- For $\lambda = -1$: solve $(A+I)v = 0$ → $v = (-2, 1)^T$
- For $\lambda = 2$: solve $(A-2I)v = 0$ → $v = (2, 1)^T$

**Step 5 — General solution:**
$$x(t) = 2C_1 e^{2t} - 2C_2 e^{-t}, \quad y(t) = C_1 e^{2t} + C_2 e^{-t}$$

**Step 6 — Check IC $(1,1)$ is not on the stable subspace** (spanned by $(-2,1)^T$):  
If it were, $C_1 = 0$ and $(x(0), y(0)) = -2C_2, C_2)$ → ratio $-2 \neq 1$. So $C_1 \neq 0$.

**Step 7 — Long-run limit:** as $t \to \infty$, $e^{-t} \to 0$:
$$\frac{x(t)}{y(t)} \to \frac{2C_1 e^{2t}}{C_1 e^{2t}} = 2$$


## 🐍 SymPy Solution

````python
t = symbols('t', real=True, positive=True)

# ── Step 1: Define the system matrix ──────────────────────────────────────
A = Matrix([[Rational(1,2), 3],
            [Rational(3,4), Rational(1,2)]])
print("Matrix A:"); pprint(A)

# ── Step 2: Eigenvalues & eigenvectors ────────────────────────────────────
print("\nEigenvalues:", A.eigenvals())
evects = A.eigenvects()
print("Eigenvectors:")
for lam, mult, vecs in evects:
    print(f"  λ={lam}: v={vecs[0].T}")

# ── Step 3: Characteristic polynomial ─────────────────────────────────────
lam = symbols('lambda')
charpoly = det(A - lam*eye(2))
print("\nCharacteristic polynomial:"); pprint(factor(charpoly))

# ── Step 4: General solution via dsolve ───────────────────────────────────
x, y = symbols('x y', cls=Function)
system = [Eq(x(t).diff(t), Rational(1,2)*x(t) + 3*y(t)),
          Eq(y(t).diff(t), Rational(3,4)*x(t) + Rational(1,2)*y(t))]
sol = dsolve(system, [x(t), y(t)])
print("\nGeneral solution:")
for s in sol:
    pprint(s)

# ── Step 5: Apply initial conditions (x(0)=1, y(0)=1) ────────────────────
C1, C2 = symbols('C1 C2')
ics_eqs = [
    sol[0].rhs.subs(t, 0) - 1,
    sol[1].rhs.subs(t, 0) - 1,
]
consts = solve(ics_eqs, [C1, C2])
print("\nConstants from IC (1,1):", consts)

x_part = sol[0].rhs.subs(consts)
y_part = sol[1].rhs.subs(consts)
print("Particular solution x(t):"); pprint(x_part)
print("Particular solution y(t):"); pprint(y_part)

# ── Step 6: Compute the limit ─────────────────────────────────────────────
ratio = x_part / y_part
lim_val = limit(ratio, t, oo)
print("\nlim_{t→∞} x(t)/y(t) =", lim_val)

# ── Step 7: Visualise ─────────────────────────────────────────────────────
x_num = lambdify(t, x_part, 'numpy')
y_num = lambdify(t, y_part, 'numpy')
ts = np.linspace(0, 3, 300)
ratio_num = x_num(ts) / y_num(ts)

plt.figure(figsize=(8, 3))
plt.plot(ts, ratio_num, color='steelblue', lw=2)
plt.axhline(2, color='tomato', ls='--', label='Limit = 2')
plt.xlabel('t'); plt.ylabel('x(t)/y(t)')
plt.title('Ratio $\\bar{x}(t)/\\bar{y}(t)$ converging to 2')
plt.legend(); plt.tight_layout(); plt.show()
````

# 📌 Problem 4 — Nonlinear ODE System: Equilibria & Stability <a id='p4'></a>


## The Problem

Given the system:
$$\dot{x} = x^2 + y^2 - 9, \qquad \dot{y} = y - x^3$$

1. How many equilibria are there?
2. Determine the stability of each equilibrium.


## 📖 Concept Recap: Nonlinear Systems — Linearisation

### Finding Equilibria
Set $\dot{x} = 0$ and $\dot{y} = 0$ simultaneously and solve.

### Jacobian Matrix
For system $\dot{x} = f(x,y)$, $\dot{y} = g(x,y)$:
$$J(x^*, y^*) = \begin{pmatrix} f_x & f_y \\ g_x & g_y \end{pmatrix}_{(x^*, y^*)}$$

The **linearised system** near equilibrium $(x^*, y^*)$ is $\dot{\xi} = J\,\xi$.

### Stability via Jacobian Eigenvalues
| Eigenvalues of $J$ | Stability |
|-------------------|----------|
| Both negative real | Stable node |
| Both positive real | Unstable node |
| Complex with $\text{Re}<0$ | Stable spiral |
| Complex with $\text{Re}>0$ | Unstable spiral |
| One positive, one negative | Saddle (unstable) |

### Shortcut via trace & determinant
- $\det J < 0$ → eigenvalues have opposite signs → **saddle** (always unstable)
- $\det J > 0$, $\text{tr}\,J < 0$ → both eigenvalues have negative real part → **stable**
- $\det J > 0$, $\text{tr}\,J > 0$ → both eigenvalues have positive real part → **unstable**

### SymPy tools
```python
equilibria = solve([f, g], [x, y])
J = Matrix([[diff(f,x), diff(f,y)], [diff(g,x), diff(g,y)]])
J.subs([(x, x_star), (y, y_star)]).eigenvals()
```


## ✏️ Hand-Solution Walkthrough

**Step 1 — Equilibrium conditions:**
$$x^2 + y^2 = 9 \quad \text{(circle of radius 3)}$$
$$y = x^3$$

**Step 2 — Count equilibria:**  
Substitute $y = x^3$: $x^2 + x^6 = 9$.  
Let $h(x) = x^6 + x^2 - 9$.  
$h(0) = -9 < 0$, $h(\pm\infty) = +\infty$, $h'(x) = 6x^5 + 2x$ has same sign as $x$.  
So $h$ is decreasing on $(-\infty,0)$ and increasing on $(0,+\infty)$ → exactly **two equilibria**: one with $x^* < 0$ (call it $a$) and one with $x^* > 0$ (call it $b$).

**Step 3 — Jacobian:**
$$J = \begin{pmatrix} 2x & 2y \\ -3x^2 & 1 \end{pmatrix}$$

**Step 4 — At equilibrium $a$ (negative $x^*$):**  
$\det J = 2x + 6x^2 y = 2x + 6x^5 = h'(x^*)/3$.  
Since $x^* < 0$: $h'(x^*) < 0$ → $\det J(a) < 0$ → **saddle** (unstable).

**Step 5 — At equilibrium $b$ (positive $x^*$):**  
$\det J(b) > 0$ and $\text{tr}\,J(b) = 1 + 2x^* > 0$ → **unstable node or spiral**.


## 🐍 SymPy Solution

````python
x, y = symbols('x y', real=True)

# ── Define the vector field ────────────────────────────────────────────────
f = x**2 + y**2 - 9
g = y - x**3

# ── Step 1: Find equilibria ────────────────────────────────────────────────
equilibria = solve([f, g], [x, y])
print(f"Number of equilibria: {len(equilibria)}")
print("Equilibria (exact):")
for eq in equilibria:
    xv = float(eq[0].evalf())
    yv = float(eq[1].evalf())
    print(f"  (x*, y*) = ({xv:.6f}, {yv:.6f})")

# ── Step 2: Jacobian ───────────────────────────────────────────────────────
J = Matrix([[diff(f,x), diff(f,y)],
            [diff(g,x), diff(g,y)]])
print("\nJacobian:"); pprint(J)

# ── Step 3: Classify each equilibrium ─────────────────────────────────────
print("\nStability analysis:")
for i, eq in enumerate(equilibria):
    x_star, y_star = eq
    J_eval = J.subs([(x, x_star), (y, y_star)])
    tr  = float(J_eval.trace().evalf())
    det = float(J_eval.det().evalf())
    eigs = J_eval.eigenvals()
    print(f"\n  Equilibrium {i+1}: ({float(x_star.evalf()):.4f}, {float(y_star.evalf()):.4f})")
    print(f"    trace = {tr:.6f}")
    print(f"    det   = {det:.6f}")
    if det < 0:
        stability = "SADDLE (unstable)"
    elif det > 0 and tr < 0:
        stability = "STABLE (node or spiral)"
    elif det > 0 and tr > 0:
        stability = "UNSTABLE (node or spiral)"
    else:
        stability = "INCONCLUSIVE"
    print(f"    → {stability}")

# ── Step 4: Phase portrait ─────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(6,6))
xs = np.linspace(-4, 4, 20)
ys = np.linspace(-4, 4, 20)
X, Y = np.meshgrid(xs, ys)
U = X**2 + Y**2 - 9
V = Y - X**3
speed = np.sqrt(U**2 + V**2) + 1e-8
ax.quiver(X, Y, U/speed, V/speed, alpha=0.4, color='gray')

theta = np.linspace(0, 2*np.pi, 300)
ax.plot(3*np.cos(theta), 3*np.sin(theta), 'b--', lw=1, label='$x^2+y^2=9$')
tx = np.linspace(-1.8, 1.8, 300)
ax.plot(tx, tx**3, 'g--', lw=1, label='$y=x^3$')

for eq in equilibria:
    xe, ye = float(eq[0].evalf()), float(eq[1].evalf())
    ax.plot(xe, ye, 'ro', ms=8)
    ax.annotate(f'({xe:.2f},{ye:.2f})', (xe,ye), textcoords="offset points",
                xytext=(6,4), fontsize=8)
ax.set_xlim(-4,4); ax.set_ylim(-4,4)
ax.set_xlabel('x'); ax.set_ylabel('y')
ax.set_title('Phase portrait — equilibria at circle ∩ cubic')
ax.legend(fontsize=8); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.show()
````

# 📌 Problem 5 — Profit Maximisation on a Non-Convex Set <a id='p5'></a>


## The Problem

A firm's feasible production set:
$$B = \{(x_1, x_2) \mid 0 \leq -e^{x_1} - x_2 + \ln(x_1^2+1) + 1\}$$

Price vector $p = (1, 1)$. The firm maximises profit $\pi = x_1 + x_2$.  
Resources are bounded below: $x_1, x_2 \geq M$ for some $M < 0$.

**Question:** Is it profitable for the firm to enter the market (produce a positive amount)?


## 📖 Concept Recap: Constrained Optimisation & SOC

### Boundary of the feasible set
At the optimum the constraint binds: $-e^{x_1} - x_2 + \ln(x_1^2+1) + 1 = 0$, i.e.
$$x_2 = -e^{x_1} + \ln(x_1^2+1) + 1$$

### Bordered Hessian / Second-Order Conditions
For maximising $f(x)$ s.t. $g(x) = 0$, the **second-order conditions** check whether the Hessian  
of the Lagrangian $D^2\mathcal{L}$ is **negative** definite on the tangent space to the constraint.

If $D^2\mathcal{L}$ is **positive** definite on the tangent space → the point is a **local minimum**, not a maximum.

### Tangent space
The tangent space to $g(x_1,x_2) = 0$ at a point is orthogonal to $\nabla g$.  
With $p = (1,1)$, the constraint gradient direction at $(0,0)$ is the direction perpendicular to $p$ → $(-1, 1)^T$.

### SymPy tools
```python
hessian(L, [x1, x2])        # Hessian matrix of L
L.diff(x1, x2)              # mixed second partial
```


## ✏️ Hand-Solution Walkthrough

**Step 1 — Reduce to boundary:**  
On the boundary: $x_2 = -e^{x_1} + \ln(x_1^2+1) + 1$.  
Profit: $\pi(x_1) = x_1 + x_2 = x_1 - e^{x_1} + \ln(x_1^2+1) + 1$.

**Step 2 — FOC:** $\pi'(x_1) = 1 + \frac{2x_1}{x_1^2+1} - e^{x_1} = 0$.  
At $x_1 = 0$: $1 + 0 - 1 = 0$ ✓ → $(0, 0)$ is a critical point.

**Step 3 — SOC (Lagrangian approach):**  
$\mathcal{L} = x_1 + x_2 - \lambda(-e^{x_1} - x_2 + \ln(x_1^2+1) + 1)$  
From $\partial\mathcal{L}/\partial x_2 = 0$: $\lambda = -1$.  

$D^2\mathcal{L}$ evaluated at $(0,0)$ with $\lambda = -1$:
$$D^2\mathcal{L}(0,0) = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$$

On the tangent direction $(-1, 1)^T$:
$$(-1,1) D^2\mathcal{L} (-1,1)^T = 1 > 0$$

**Conclusion:** $(0,0)$ is a **local minimum**, not a maximum.  
Since the feasible set is bounded and $(0,0)$ is a strict local min, the maximum must yield positive profit → **yes, it is profitable to enter the market**.


## 🐍 SymPy Solution

````python
x1, x2, lam = symbols('x1 x2 lambda', real=True)

# ── Step 1: Define constraint and Lagrangian ───────────────────────────────
constraint = -exp(x1) - x2 + log(x1**2 + 1) + 1
profit     = x1 + x2
L = profit - lam * constraint

print("Lagrangian L:"); pprint(L)

# ── Step 2: First-order conditions ────────────────────────────────────────
dL_x1  = diff(L, x1)
dL_x2  = diff(L, x2)
dL_lam = diff(L, lam)   # = constraint = 0

print("\nFOC ∂L/∂x1:"); pprint(dL_x1)
print("FOC ∂L/∂x2:"); pprint(dL_x2)

# Solve lambda from ∂L/∂x2 = 0
lam_val = solve(dL_x2, lam)[0]
print("\nλ =", lam_val)

# Condition on x1 from ∂L/∂x1 = 0 with λ substituted
h = dL_x1.subs(lam, lam_val)
print("\nCondition h(x1) = 0:"); pprint(h)
print("h(0) =", h.subs(x1, 0))

# Solve for x2 from constraint at x1=0
x2_val = solve(constraint.subs(x1, 0), x2)[0]
print(f"\nAt x1=0: x2 = {x2_val}")
print("→ Critical point: (0, 0)")

# ── Step 3: Second-order conditions ───────────────────────────────────────
# Hessian of L
D2L = Matrix([[diff(L, x1, x1), diff(L, x1, x2)],
              [diff(L, x2, x1), diff(L, x2, x2)]])
D2L_at_critical = D2L.subs([(x1, 0), (x2, 0), (lam, lam_val)])
print("\nD²L at (0,0):"); pprint(D2L_at_critical)

# Tangent vector (perpendicular to ∇f = (1,1) → tangent = (-1,1))
tangent = Matrix([-1, 1])
soc_val = (tangent.T * D2L_at_critical * tangent)[0]
print(f"\nSOC check: tangentᵀ D²L tangent = {soc_val}")
print("→ Positive → (0,0) is a LOCAL MINIMUM, not a maximum!")

# ── Step 4: Show profit is positive near (0,0) ────────────────────────────
# Parameterise the boundary: x2 = -exp(x1) + log(x1²+1) + 1
x2_boundary = -exp(x1) + log(x1**2 + 1) + 1
profit_on_boundary = x1 + x2_boundary
print("\nProfit along boundary π(x1) = x1 + x2_boundary:")
pprint(profit_on_boundary)
print("π(0) =", profit_on_boundary.subs(x1, 0))
print("π(0.5) ≈", float(profit_on_boundary.subs(x1, 0.5).evalf()))
print("π(1.0) ≈", float(profit_on_boundary.subs(x1, 1.0).evalf()))
print("π(-0.5) ≈", float(profit_on_boundary.subs(x1, -0.5).evalf()))

# ── Step 5: Plot profit along boundary ────────────────────────────────────
pi_num = lambdify(x1, profit_on_boundary, 'numpy')
x1_vals = np.linspace(-2, 2, 400)

# Keep only feasible points
x2_vals = -np.exp(x1_vals) + np.log(x1_vals**2 + 1) + 1

plt.figure(figsize=(9, 4))
plt.subplot(1, 2, 1)
plt.plot(x1_vals, pi_num(x1_vals), 'steelblue', lw=2)
plt.axhline(0, color='k', lw=0.8)
plt.axvline(0, color='k', lw=0.8)
plt.scatter([0], [0], color='red', zorder=5, label='Local min at (0,0)')
plt.xlabel('$x_1$'); plt.ylabel('Profit $\\pi$')
plt.title('Profit along boundary')
plt.legend(); plt.grid(True, alpha=0.3)

plt.subplot(1, 2, 2)
plt.contourf(np.meshgrid(x1_vals, np.linspace(-3, 2, 100))[0],
             np.meshgrid(x1_vals, np.linspace(-3, 2, 100))[1],
             np.meshgrid(x1_vals, np.linspace(-3, 2, 100))[0] + np.meshgrid(x1_vals, np.linspace(-3, 2, 100))[1],
             levels=20, cmap='RdYlGn', alpha=0.7)
plt.plot(x1_vals, x2_vals, 'k-', lw=2, label='Boundary of B')
plt.scatter([0], [0], color='red', zorder=5, s=80, label='(0,0)')
plt.colorbar(label='Profit')
plt.xlabel('$x_1$'); plt.ylabel('$x_2$'); plt.title('Production set B')
plt.xlim(-2, 2); plt.ylim(-3, 2); plt.legend(fontsize=8)
plt.tight_layout(); plt.show()
````
