## 🔷 1 — Symbols, Expressions & Printing <a id='s1'></a>


### 📖 Recap

**Declaring symbols** — the first thing you do in every SymPy problem:
```python
t = symbols('t')                          # generic real
t = symbols('t', real=True)              # constrained: t ∈ ℝ
t = symbols('t', real=True, positive=True) # t > 0
x, y = symbols('x y', real=True)         # multiple at once
x1, x2, x3, lam, w = symbols('x1 x2 x3 lambda w', positive=True)
```

**`Function`** — for unknown functions of a variable (used in ODEs):
```python
x = Function('x')    # x is now a function, called as x(t)
x(t)                 # the expression x(t)
x(t).diff(t)         # ẋ(t) = dx/dt
```

**`Rational`** — exact fractions (avoids floating-point errors):
```python
Rational(1, 2)       # ½  exactly
Rational(3, 4)       # ¾
# Used in the exam: Matrix([[Rational(1,2), 3], [Rational(3,4), Rational(1,2)]])
```

**Printing:**
```python
pprint(expr)          # pretty-print in terminal
display(expr)         # rendered LaTeX in Jupyter
print(expr)           # plain text  x**2 + 1
```

**`Eq`** — write explicit equations:
```python
Eq(x(t).diff(t), x(t))     # ẋ = x
Eq(lhs, rhs)               # lhs = rhs
```


### 🔍 Demo

````python
# Symbols used in the exam
t = symbols('t', real=True)
x1, x2, x3 = symbols('x1 x2 x3', positive=True)
p1, p2, p3, lam, w = symbols('p1 p2 p3 lambda w', positive=True)
a1, a2, a3 = Rational(1,2), Rational(1,3), Rational(1,6)

print("Exact fractions:", a1, "+", a2, "+", a3, "=", a1+a2+a3)

# Function for ODE
x_fn = Function('x')
expr = (t**2 - 1) * x_fn(t)
print("\nExpression:", expr)
print("Derivative: ", expr.diff(t))

# Eq for an ODE
ode = Eq(x_fn(t).diff(t), x_fn(t))
print("\nODE:"); pprint(ode)

# Rational matrix entries
A = Matrix([[Rational(1,2), 3], [Rational(3,4), Rational(1,2)]])
print("\nExact matrix A:"); pprint(A)
print("Compare float A:"); pprint(A.evalf())
````

## 🔷 2 — Expand, Factor, Simplify, Cancel <a id='s2'></a>


### 📖 Recap

| Function | What it does | When to use |
|----------|-------------|-------------|
| `expand(expr)` | Multiply out all brackets | After substitution, before solving |
| `factor(expr)` | Factorise completely | Characteristic polynomials |
| `simplify(expr)` | General simplification (slow) | Final answer cleanup |
| `cancel(expr)` | Cancel common rational factors | Simplify `(t²-1)/(t-1)` |
| `trigsimp(expr)` | Trig identities | `sin²+cos²→1` |

```python
t = symbols('t')

expand((t-2)*t - 1)           # t² - 2t - 1
factor(t**2 - 2*t - 1)        # not factorable over ℤ
factor(t**2 - 1)               # (t-1)(t+1)
simplify((t**2-1)/(t-1))       # t+1
cancel((t**2-1)/(t-1))         # t+1   (faster than simplify)
```

**Used in Problem 1** to spot the substitution:
```python
expand(((t-2)*t - 1))          # t² - 2t - 1
# rearranged: (t²-1) - 2t  →  substitution y = (t²-1)x
```

**Used in Problem 3** for the characteristic polynomial:
```python
factor(lam**2 - lam - 2)       # (λ-2)(λ+1)
```


### 🔍 Demo

````python
t, lam = symbols('t lambda')

# Problem 1: expand the LHS coefficient
coeff = (t - 2)*t - 1
print("Expanded LHS coeff:", expand(coeff))
print("= (t²-1) - 2t:", expand(coeff - ((t**2-1) - 2*t)), "== 0?")

# Problem 3: characteristic polynomial
A = Matrix([[Rational(1,2), 3],[Rational(3,4), Rational(1,2)]])
charpoly = det(A - lam*eye(2))
print("\nChar poly:", charpoly)
print("Factored:  ", factor(charpoly))

# cancel and simplify
expr = (t**2 - 1) / (t - 1)
print("\ncancel((t²-1)/(t-1)):  ", cancel(expr))
print("simplify((t²-1)/(t-1)):", simplify(expr))

# expand after substitution
x = symbols('x')
expr2 = (x + 1)**3
print("\nexpand((x+1)³):", expand(expr2))
````

## 🔷 3 — Substitution (.subs) & Evaluation (.evalf) <a id='s3'></a>


### 📖 Recap

**`.subs(var, value)`** — substitute a value or expression:
```python
expr.subs(x, 5)                   # substitute x=5
expr.subs([(x, 2), (y, 3)])       # multiple at once — list of tuples
J.subs([(x, x_star), (y, y_star)]) # evaluate Jacobian at equilibrium
```

**`.evalf(n)`** — evaluate to n significant digits:
```python
sqrt(2).evalf()        # 1.41421356237310  (15 digits, default)
sqrt(2).evalf(30)      # 30 digits
pi.evalf(50)
```

**`float(expr)`** — convert SymPy number to Python float:
```python
float(equilibria[0][0].evalf())    # extract numerical x* as Python float
```

**`lambdify`** — convert expression to fast numerical function:
```python
from sympy import lambdify
import numpy as np
f_num = lambdify(t, expr, 'numpy')    # vectorised over numpy arrays
f_num(np.linspace(0, 5, 100))
```

## 🔷 4 — Differentiation: diff & partial derivatives <a id='s4'></a>


### 📖 Recap

```python
diff(expr, x)        # first derivative w.r.t. x
diff(expr, x, 2)     # second derivative
diff(expr, x, y)     # mixed partial ∂²f/∂x∂y
expr.diff(x)         # method syntax — identical
```

**Building the Jacobian manually** (Problem 4):
```python
f = x**2 + y**2 - 9
g = y - x**3
J = Matrix([[diff(f,x), diff(f,y)],
            [diff(g,x), diff(g,y)]])
```

**Building FOCs** (Problems 2, 5):
```python
L = log_u - lam*(p1*x1 + p2*x2 - w)
focs = [diff(L, x1), diff(L, x2), diff(L, lam)]
```

**Hessian of Lagrangian** (Problem 5):
```python
D2L = Matrix([[diff(L, x1, x1), diff(L, x1, x2)],
              [diff(L, x2, x1), diff(L, x2, x2)]])
# OR use SymPy's built-in:
H = hessian(L, [x1, x2])
```

**Second-order derivative for 1-variable problems**:
```python
pi = x1 - exp(x1) + log(x1**2 + 1) + 1
dpi  = diff(pi, x1)         # first derivative
d2pi = diff(pi, x1, 2)      # second derivative — for classification
```


### 🔍 Demo

````python
x, y, x1, x2, lam, p1, p2, w = symbols('x y x1 x2 lambda p1 p2 w', positive=True)
a1, a2 = Rational(3,5), Rational(2,5)

# ── Problem 2: Lagrangian and FOCs ────────────────────────────────────────
L = a1*log(x1) + a2*log(x2) - lam*(p1*x1 + p2*x2 - w)
print("Lagrangian L ="); pprint(L)
focs = [diff(L, v) for v in [x1, x2, lam]]
print("\nFOCs:")
for foc in focs:
    pprint(foc)

# ── Problem 4: Jacobian ────────────────────────────────────────────────────
print("\n── Jacobian ──")
f = x**2 + y**2 - 9
g = y - x**3
J = Matrix([[diff(f,x), diff(f,y)],
            [diff(g,x), diff(g,y)]])
pprint(J)

# ── Problem 5: Hessian of L ────────────────────────────────────────────────
print("\n── Hessian of L (Problem 5) ──")
x1r, x2r, lamr = symbols('x1 x2 lambda', real=True)
constraint = -exp(x1r) - x2r + log(x1r**2 + 1) + 1
Lp5 = x1r + x2r - lamr * constraint
D2L = hessian(Lp5, [x1r, x2r])
print("D²L:"); pprint(D2L)
print("D²L at (0,0):"); pprint(D2L.subs([(x1r,0),(x2r,0),(lamr,-1)]))
````

## 🔷 5 — Solving Equations: solve, solveset, nsolve <a id='s5'></a>


### 📖 Recap

**`solve`** — the main workhorse:
```python
solve(expr, x)                        # solve expr=0 for x → list
solve([eq1, eq2], [x, y])             # system → dict  {x:val, y:val}
solve(focs, [x1,x2,x3,lam], dict=True)[0]  # dict form, exam pattern
```

> ⚠️ `solve` returns a **list**. For a system, get the first solution with `[0]`.  
> Always check: `if sol:` before accessing `sol[0]`.

**`solveset`** — modern version, returns a set:
```python
solveset(x**2 - 4, x, S.Reals)       # FiniteSet(-2, 2)
solveset(x**2 + 1, x, S.Reals)       # EmptySet — no real solutions
```

**`nsolve`** — numerical solver, needs an initial guess:
```python
nsolve(expr, x, 0.5)                  # find root near x=0.5
nsolve([f, g], [x, y], [1, 1])        # system, initial guess (1,1)
```

**Exam patterns:**
```python
# Problem 2: solve 4-equation FOC system
sol = solve(focs, [x1, x2, x3, lam], dict=True)[0]

# Problem 4: find equilibria
equilibria = solve([f, g], [x, y])   # returns list of (x,y) tuples

# Problem 5: solve for lambda from one FOC
lam_val = solve(diff(L, x2), lam)[0]

# Problem 5: nsolve for transcendental equation
cp = nsolve(diff(pi, x1), x1, 0.5)
```


### 🔍 Demo

````python
x1, x2, lam, p1, p2, w = symbols('x1 x2 lambda p1 p2 w', positive=True)
a1, a2 = Rational(3,5), Rational(2,5)

# ── Problem 2: solve FOC system ───────────────────────────────────────────
L = a1*log(x1) + a2*log(x2) - lam*(p1*x1 + p2*x2 - w)
focs = [diff(L, x1), diff(L, x2), diff(L, lam)]
sol = solve(focs, [x1, x2, lam], dict=True)[0]
print("Marshallian demands:")
for var, expr in sol.items():
    print(f"  {var}* =", simplify(expr))

# ── Problem 5: extract lambda ─────────────────────────────────────────────
x1r, x2r, lamr = symbols('x1 x2 lambda', real=True)
constraint = -exp(x1r) - x2r + log(x1r**2 + 1) + 1
Lp5 = x1r + x2r - lamr * constraint
lam_val = solve(diff(Lp5, x2r), lamr)[0]
print("\nλ from ∂L/∂x2=0:", lam_val)

# ── Problem 4: find equilibria numerically ────────────────────────────────
x, y = symbols('x y', real=True)
f = x**2 + y**2 - 9
g = y - x**3
print("\nEquilibria (approximate):")
equilibria = solve([f, g], [x, y])
for eq in equilibria:
    print(f"  ({float(eq[0].evalf()):.6f}, {float(eq[1].evalf()):.6f})")

# ── nsolve for transcendental FOC ─────────────────────────────────────────
pi = x1r - exp(x1r) + log(x1r**2 + 1) + 1
dpi = diff(pi, x1r)
print("\nπ'(x1):"); pprint(dpi)
for guess in [-1.0, 0.0, 1.0]:
    try:
        cp = nsolve(dpi, x1r, guess)
        print(f"  nsolve(guess={guess:+.1f}): x1* = {float(cp):.6f}, π''= {float(diff(pi,x1r,2).subs(x1r,cp)):.4f}")
    except Exception as e:
        print(f"  nsolve(guess={guess:+.1f}): failed — {e}")
````

## 🔷 6 — Solving ODEs: dsolve, Function, Eq <a id='s6'></a>


### 📖 Recap

**Step 1 — Declare the unknown function:**
```python
t = symbols('t')
x = Function('x')      # x is now a function, used as x(t)
```

**Step 2 — Write the ODE as an `Eq`:**
```python
# Single ODE
ode = Eq(x(t).diff(t) - x(t), 0)        # ẋ - x = 0

# System of ODEs (list)
system = [Eq(x(t).diff(t), a*x(t) + b*y(t)),
          Eq(y(t).diff(t), c*x(t) + d*y(t))]
```

**Step 3 — Solve:**
```python
sol = dsolve(ode, x(t))        # single ODE → Eq(x(t), C1*exp(t))
sol = dsolve(system, [x(t), y(t)])  # system → list of Eq
```

**Step 4 — Apply initial conditions:**
```python
C1, C2 = symbols('C1 C2')
# Extract RHS of each solution
x_rhs = sol[0].rhs
y_rhs = sol[1].rhs
# Evaluate at t=0
ics = [x_rhs.subs(t, 0) - x0,   # = 0
       y_rhs.subs(t, 0) - y0]    # = 0
consts = solve(ics, [C1, C2])
# Substitute constants back
x_particular = x_rhs.subs(consts)
```

**Exam patterns:**
- Problem 1: single ODE, substitution `y = (t²-1)x` reduces it to `ẏ = y`
- Problem 3: linear 2×2 system, then ICs, then limit of ratio


### 🔍 Demo

````python
t = symbols('t', real=True)
x, y = symbols('x y', cls=Function)
C1, C2 = symbols('C1 C2')

# ── Problem 1: Solve the reduced ODE ẏ = y ────────────────────────────────
ode_simple = Eq(y(t).diff(t), y(t))
sol_y = dsolve(ode_simple, y(t))
print("Solution of ẏ = y:"); pprint(sol_y)

# Undo substitution y = (t²-1)*x  →  x = y/(t²-1)
x_from_y = sol_y.rhs / (t**2 - 1)
print("\nGeneral solution x(t):"); pprint(simplify(x_from_y))

# ── Problem 3: Linear system ───────────────────────────────────────────────
print("\n── Problem 3: System ──")
system = [
    Eq(x(t).diff(t), Rational(1,2)*x(t) + 3*y(t)),
    Eq(y(t).diff(t), Rational(3,4)*x(t) + Rational(1,2)*y(t))
]
sol = dsolve(system, [x(t), y(t)])
print("General solution:")
for s in sol: pprint(s)

# Apply IC (1,1)
ics_eqs = [sol[0].rhs.subs(t,0) - 1,
           sol[1].rhs.subs(t,0) - 1]
consts = solve(ics_eqs, [C1, C2])
print("\nConstants:", consts)
x_part = sol[0].rhs.subs(consts)
y_part = sol[1].rhs.subs(consts)
print("x(t) ="); pprint(x_part)
print("y(t) ="); pprint(y_part)
````

## 🔷 7 — Matrices: Matrix, det, trace, inv, eigenvals <a id='s7'></a>


### 📖 Recap

```python
A = Matrix([[a, b], [c, d]])     # create
eye(2)                           # 2×2 identity
zeros(2, 2)                      # zero matrix

A.det()          # determinant
A.trace()        # sum of diagonal = sum of eigenvalues
A.inv()          # inverse  (raises error if det=0)
A.T              # transpose
A * B            # matrix multiplication
A.eigenvals()    # {λ: multiplicity, ...}
A.eigenvects()   # [(λ, mult, [v1, ...]), ...]
A.charpoly()     # characteristic polynomial as Poly object
det(A - lam*eye(2))  # characteristic polynomial as expression
A.diagonalize()  # (P, D)  where A = P D P^{-1}
```

**Exam patterns:**

```python
# Problem 3: build and analyse A

print(A.eigenvals())          # {2: 1, -1: 1}

# Problem 4: Jacobian at equilibrium
J_eval = J.subs([(x, x_star), (y, y_star)])
J_eval.trace()    # classification shortcut
J_eval.det()
```

## 🔷 9 — Hessian & Second-Order Conditions <a id='s9'></a>


### 📖 Recap

**`hessian(f, vars)`** — builds the Hessian matrix automatically:
```python
H = hessian(f, [x1, x2])
# equivalent to:
H = Matrix([[diff(f,x1,x1), diff(f,x1,x2)],
            [diff(f,x2,x1), diff(f,x2,x2)]])
```

**Checking definiteness** (2×2):
| Condition | Type |
|-----------|------|
| `H[0,0] < 0` and `det(H) > 0` | Negative definite → **local max** |
| `H[0,0] > 0` and `det(H) > 0` | Positive definite → **local min** |
| `det(H) < 0` | Indefinite → saddle point |

**Tangent-space SOC** (constrained problems):
```python
# Tangent to constraint g=0 at the point, perpendicular to ∇f = (1,1):
tangent = Matrix([-1, 1])
soc = (tangent.T * D2L * tangent)[0]  # scalar
# soc > 0 → local min; soc < 0 → local max
```

**Bordered Hessian** (for constrained optimisation):
```python
g = constraint_expr
Hbar = Matrix([
    [0,           diff(g,x1),       diff(g,x2)      ],
    [diff(g,x1),  diff(L,x1,x1),    diff(L,x1,x2)   ],
    [diff(g,x2),  diff(L,x2,x1),    diff(L,x2,x2)   ]
])
# For maximum: det(Hbar) > 0
```

**Exam use — Problem 5:**  
At $(0,0)$ with $\lambda=-1$: $D^2\mathcal{L}(0,0) = \begin{pmatrix}1&0\\0&0\end{pmatrix}$.  
Tangent direction $(-1,1)^T$: $(-1,1)\cdot D^2\mathcal{L}\cdot(-1,1)^T = 1 > 0$ → **local min**.


### 🔍 Demo

````python
x1, x2, lam = symbols('x1 x2 lambda', real=True)

# ── Problem 5: full SOC analysis ──────────────────────────────────────────
constraint = -exp(x1) - x2 + log(x1**2 + 1) + 1
L = x1 + x2 - lam * constraint

print("Lagrangian:"); pprint(L)

# Hessian of L
D2L = hessian(L, [x1, x2])
print("\nD²L ="); pprint(D2L)

# Evaluate at critical point (0,0) with λ=-1
D2L_at_cp = D2L.subs([(x1,0),(x2,0),(lam,-1)])
print("\nD²L at (0,0):"); pprint(D2L_at_cp)
print("D²L[0,0] =", D2L_at_cp[0,0], "→", "< 0 → could be max" if D2L_at_cp[0,0] < 0 else "> 0 → local min direction")
print("det(D²L) =", D2L_at_cp.det())

# Tangent-space check
tangent = Matrix([-1, 1])
soc = (tangent.T * D2L_at_cp * tangent)[0]
print(f"\nSOC (tangent direction): {soc} →", "local MIN" if soc > 0 else "local MAX")

# Bordered Hessian
g = constraint
Hbar = Matrix([
    [0,          diff(g,x1),       diff(g,x2)      ],
    [diff(g,x1), diff(L,x1,x1),    diff(L,x1,x2)   ],
    [diff(g,x2), diff(L,x2,x1),    diff(L,x2,x2)   ]
]).subs([(x1,0),(x2,0),(lam,-1)])
print("\nBordered Hessian:"); pprint(Hbar)
print("det(H̄) =", Hbar.det(), "→", "MAX" if Hbar.det() > 0 else "MIN / not MAX")
````
