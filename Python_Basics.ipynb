## 🔷 2 — f-strings & Formatted Output <a id='s2'></a>


### 📖 Recap

f-strings embed expressions directly inside string literals:

```python
name = "Alice"
score = 95.678
print(f"Name: {name}, Score: {score:.2f}")   # 2 decimal places
```

| Spec | Meaning | Example output |
|------|---------|----------------|
| `:.4f` | 4 decimal places | `3.1416` |
| `:.6f` | 6 decimal places | `3.141593` |
| `:.4e` | scientific notation | `3.1416e+00` |
| `:>10` | right-align in 10 chars | `    3.1416` |
| `:<10` | left-align | `3.1416    ` |
| `:^12` | centre | `  3.1416   ` |
| `:+.4f` | always show sign | `+3.1416` |

**Used in the exam to:**
- Print equilibrium coordinates: `f"(x*, y*) = ({xv:.6f}, {yv:.6f})"`
- Print stability: `f"  trace = {tr:.6f}"`
- Build summary tables over a loop

````python
# Basic f-string formatting
lam1, lam2 = 2.0, -1.0
print(f"Eigenvalues: λ₁ = {lam1:.4f}, λ₂ = {lam2:.4f}")

# Table header + rows
print(f"\n{'Equilibrium':<15} {'x*':>10} {'y*':>10} {'trace':>10} {'det':>10} {'type':>12}")
print("-" * 60)
rows = [("a (x<0)", -1.3216, -2.3083, -1.6432, -0.3, "saddle"),
        ("b (x>0)",  1.3216,  2.3083,  3.6432,  1.8, "unstable")]
for name, xv, yv, tr, de, typ in rows:
    print(f"{name:<15} {xv:>10.4f} {yv:>10.4f} {tr:>10.4f} {de:>10.4f} {typ:>12}")

# Sign formatting
for val in [-2.0, 0.0, 2.0]:
    print(f"  {val:+.2f}")
````

```python
if det < 0:
```

**Combining conditions:**
```python

not   # invert
```


### 🔍 Demo

````python
def classify_equilibrium(trace, det):
    """Classify a 2D linear system equilibrium from trace and determinant."""
    if det < 0:
        return "saddle (always unstable)"
    elif det == 0:
        return "degenerate (non-isolated equilibrium)"
    elif trace < 0:
        return "stable node or spiral"
    elif trace > 0:
        return "unstable node or spiral"
    else:
        return "centre (neutrally stable)"

test_cases = [(-1, 2), (1, 3), (-2, -1), (0, 4), (2, -0.5)]
for tr, de in test_cases:
    print(f"  trace={tr:+.1f}, det={de:+.1f}  →  {classify_equilibrium(tr, de)}")
````

### 📖 Recap


**Multiple return values** (returns a tuple):
```python
def eigenvalues_2x2(a, b, c, d):
    tr   = a + d
    det  = a*d - b*c
    disc = tr**2 - 4*det
    return tr, det, disc     # unpack: t, d, di = eigenvalues_2x2(...)
```

**Default arguments:**
```python
def classify(trace, det, tol=1e-10):
    if abs(det) < tol:
        return "degenerate"
    ...
```

**Used in the exam** to:
- Wrap the Jacobian evaluation: `def jac_at(x, y): ...`
- Wrap stability classification: `def classify(tr, det): ...`
- Wrap the profit-along-boundary: `def profit(x1): ...`


### 🔍 Demo

````python
import math

def eigenvalues_2x2(a, b, c, d):
    """Return eigenvalues of [[a,b],[c,d]] using the quadratic formula."""
    tr   = a + d
    det  = a*d - b*c
    disc = tr**2 - 4*det
    if disc >= 0:
        lam1 = (tr + math.sqrt(disc)) / 2
        lam2 = (tr - math.sqrt(disc)) / 2
        return lam1, lam2
    else:
        real = tr / 2
        imag = math.sqrt(-disc) / 2
        return complex(real, imag), complex(real, -imag)

def classify(trace, det):
    """Classify a 2D equilibrium."""
    if det < 0:     return "saddle"
    elif trace < 0: return "stable"
    elif trace > 0: return "unstable"
    else:           return "centre"

# Exam matrix A = [[1/2, 3],[3/4, 1/2]]
l1, l2 = eigenvalues_2x2(0.5, 3, 0.75, 0.5)
print(f"λ₁ = {l1}, λ₂ = {l2}")
print("Classification:", classify(0.5+0.5, 0.5*0.5 - 3*0.75))
````


````python
import math

# Safe square root
def safe_sqrt(x):
    try:
        if x < 0:
            raise ValueError(f"Cannot take sqrt of {x}")
        return math.sqrt(x)
    except ValueError as e:
        print(f"  Warning: {e}")
        return None

for val in [4, -1, 9, -3]:
    result = safe_sqrt(val)
    print(f"  sqrt({val}) = {result}")

# Guard against empty solution list
def safe_solve_example(discriminant):
    """Returns eigenvalues or warns if no real solution."""
    try:
        if discriminant < 0:
            raise ValueError("Complex eigenvalues")
        lam1 = (1 + math.sqrt(discriminant)) / 2
        lam2 = (1 - math.sqrt(discriminant)) / 2
        return [lam1, lam2]
    except ValueError as e:
        print(f"  Note: {e}")
        return []

print()
for disc in [9, 0, -4]:
    sols = safe_solve_example(disc)
    print(f"  disc={disc:+d}: solutions = {[round(s,3) for s in sols]}")
````
